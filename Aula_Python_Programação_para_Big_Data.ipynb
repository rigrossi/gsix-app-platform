```python
import math

# Numeros Primos

numero = int(input("Digite um numero: "))

# Assume que o numero é primo por padrão
primo = True

if numero < 2:
  primo = False
else:
  # Itera de 2 até a raiz quadrada do numero (incluindo a raiz quadrada)
  # Adicionamos 1 ao limite superior para incluir o valor da raiz quadrada na iteração
  for i in range(2, int(math.sqrt(numero)) + 1):
    if numero % i == 0:
      primo = False
      break

if primo:
  print(f"{numero} é um número primo")
else:
  print(f"{numero} não é um número primo")
```

```python
while True:
  entrada = input("Digite o valor do produto ou 'Fim' para encerrar o programa... ")

  if entrada.upper() == "Fim":
    print("Processo finalizado pelo usuario...")
    break

  try:
    valor_produto = float(entrada)
  except ValueError:
    print("valor inválido, informe numeros...")
    continue

  codigo_produto = input("Informe o codigo do produto...")

  desconto = 0.0
  if valor_produto <= 100.0:
    desconto = 0.05
  elif valor_produto <= 250.0:
    desconto = 0.08
  elif valor_produto <= 450.0:
    desconto = 0.1
  else:
    desconto = 0.15

  preco_total = valor_produto - (valor_produto * desconto)

  print(f"Codigo do produto: {codigo_produto}")
  print(f"Valor do produto: {valor_produto}")
  print(f"Valor total: {preco_total}")

print("Linha final do programa......")
```

```python
# Passo 1: Ler os livros existentes do arquivo
try:
    with open('livros.txt', 'r') as arquivo: # w write, r read
        livros = [linha.strip() for linha in arquivo.readlines()]
except FileNotFoundError:
    livros = []  # Caso o arquivo ainda não exista

# Transforma em conjunto para evitar duplicatas
livros = list(set(livros))

print(livros)

print("Digite os títulos dos livros que você leu (digite 'fim' para encerrar):")

# Passo 2: Adicionar novos livros lidos
while True:
    novo_livro = input("Informe o Livro lido: ").strip()

    if novo_livro.lower() == 'fim':
        break

    if novo_livro not in livros:
        livros.append(novo_livro)
    else:
        print("Esse livro já está na lista.")


# Passo 3: Ordenar a lista por ordem alfabética
livros.sort()

# Passo 4: Atualizar o arquivo com os livros únicos e ordenados
with open('livros.txt', 'w') as arquivo:
    for livro in livros:
        arquivo.write(livro + '\n')

# Passo 5: Mostrar a lista atualizada
print("\n📚 Lista atualizada de livros lidos:")
for livro in livros:
    print("- " + livro)
```

```python
import pandas as pd

# Criar um DataFrame de exemplo
data = {
    "Categoria": ["A", "A", "B", "B", "C", "C"],
    "Valor":     [10 ,  20,  30,  40,  50,  60]
}

df = pd.DataFrame(data)

# Agrupar os dados pela coluna 'Categoria' e calcular a soma dos valores em cada grupo
grupo = df.groupby("Categoria")["Valor"].sum().reset_index()

print(grupo)
```

```python
import pandas as pd

# Criar o DataFrame com os dados fornecidos
data = {
    "Data": ["2023-01-01", "2023-01-01", "2023-01-02", "2023-01-03", "2023-01-04"],
    "Produto": ["Arroz", "Feijão", "Macarrão", "Leite", "Pão"],
    "Categoria": ["Grãos", "Grãos", "Massas", "Laticínios", "Padaria"],
    "Quantidade": [2, 1, 3, 5, 6],
    "Preco_Unitario": [5.00, 4.00, 3.00, 2.00, 1.00],
    "Total_Compra": [10.00, 4.00, 9.00, 10.00, 6.00]
}

df = pd.DataFrame(data)

# Tabela dinâmica que lista os produtos e suas quantidades totais vendidas
pivot_produtos = df.groupby("Produto")["Quantidade"].sum().reset_index()

#print(pivot_produtos)

# Tabela dinâmica que soma os valores da coluna Total_Compra para cada categoria
pivot_categorias = df.groupby("Categoria")["Total_Compra"].sum().reset_index()

print(pivot_categorias)
```
